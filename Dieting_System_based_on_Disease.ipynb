```python
# Dieting System based on Disease

diet = {
    "diabetes": ["vegetables", "whole grains", "nuts", "fruits"],
    "high_bp": ["vegetables", "fruits", "oats", "low salt food"],
    "obesity": ["vegetables", "salads", "fruits", "lean protein"],
    "anemia": ["spinach", "beans", "meat", "eggs"],
    "heart_disease": ["oats", "fish", "vegetables", "fruits"]
}

disease = input("Enter disease: ").lower()

if disease in diet:
    print("Recommended foods:")
    for food in diet[disease]:
        print("-", food)
else:
    print("Disease not found.")
```
