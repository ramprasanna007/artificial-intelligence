TIC-TAC-TOE

```python
"""
Tic-Tac-Toe: Human vs AI
------------------------
You play X, the AI plays O.
The AI uses minimax with alpha-beta pruning, so on "hard" it never loses.

Run:
    python tic_tac_toe.py
"""

import math
import random

HUMAN = "X"
AI = "O"
EMPTY = " "

WIN_LINES = [
    (0, 1, 2), (3, 4, 5), (6, 7, 8),   # rows
    (0, 3, 6), (1, 4, 7), (2, 5, 8),   # columns
    (0, 4, 8), (2, 4, 6),              # diagonals
]


def new_board():
    return [EMPTY] * 9


def print_board(board):
    rows = [board[i:i + 3] for i in range(0, 9, 3)]
    print()
    for r, row in enumerate(rows):
        print(" " + " | ".join(row))
        if r < 2:
            print("---+---+---")
    print()


def winner(board):
    for a, b, c in WIN_LINES:
        if board[a] != EMPTY and board[a] == board[b] == board[c]:
            return board[a]
    return None


def is_full(board):
    return EMPTY not in board


def empty_cells(board):
    return [i for i, v in enumerate(board) if v == EMPTY]


def game_over(board):
    return winner(board) is not None or is_full(board)


def minimax(board, depth, is_maximizing, alpha, beta):
    win = winner(board)
    if win == AI:
        return 10 - depth
    if win == HUMAN:
        return depth - 10
    if is_full(board):
        return 0

    if is_maximizing:
        best = -math.inf
        for i in empty_cells(board):
            board[i] = AI
            best = max(best, minimax(board, depth + 1, False, alpha, beta))
            board[i] = EMPTY
            alpha = max(alpha, best)
            if beta <= alpha:
                break
        return best
    else:
        best = math.inf
        for i in empty_cells(board):
            board[i] = HUMAN
            best = min(best, minimax(board, depth + 1, True, alpha, beta))
            board[i] = EMPTY
            beta = min(beta, best)
            if beta <= alpha:
                break
        return best


def best_move(board):
    best_score = -math.inf
    move = None
    for i in empty_cells(board):
        board[i] = AI
        score = minimax(board, 0, False, -math.inf, math.inf)
        board[i] = EMPTY
        if score > best_score:
            best_score = score
            move = i
    return move


def easy_move(board):
    # 60% random, 40% optimal -> beatable but not clueless
    if random.random() < 0.6:
        return random.choice(empty_cells(board))
    return best_move(board)


def ask_human_move(board):
    while True:
        raw = input("Your move (1-9): ").strip()
        if not raw.isdigit():
            print("Please enter a number from 1 to 9.")
            continue
        idx = int(raw) - 1
        if idx < 0 or idx > 8:
            print("Please enter a number from 1 to 9.")
            continue
        if board[idx] != EMPTY:
            print("That square is taken. Try again.")
            continue
        return idx


def choose_difficulty():
    print("Choose difficulty:")
    print("  1) Easy (beatable)")
    print("  2) Unbeatable")
    while True:
        choice = input("Enter 1 or 2: ").strip()
        if choice == "1":
            return "easy"
        if choice == "2":
            return "hard"
        print("Please enter 1 or 2.")


def play_round(difficulty):
    board = new_board()
    human_turn = random.choice([True, False])
    print("\nSquares are numbered 1-9, left to right, top to bottom.")
    print_board(["1", "2", "3", "4", "5", "6", "7", "8", "9"])
    print("You are X. AI is O.")
    print("You go first!" if human_turn else "AI goes first!")

    while not game_over(board):
        if human_turn:
            idx = ask_human_move(board)
            board[idx] = HUMAN
        else:
            print("AI is thinking...")
            idx = easy_move(board) if difficulty == "easy" else best_move(board)
            board[idx] = AI
        print_board(board)
        human_turn = not human_turn

    win = winner(board)
    if win == HUMAN:
        print("You win!")
        return "human"
    elif win == AI:
        print("AI wins!")
        return "ai"
    else:
        print("It's a draw.")
        return "draw"


def main():
    print("=== Tic-Tac-Toe: Human vs AI ===")
    difficulty = choose_difficulty()
    scores = {"human": 0, "ai": 0, "draw": 0}

    while True:
        result = play_round(difficulty)
        scores[result] += 1
        print(f"\nScore — You: {scores['human']}  AI: {scores['ai']}  Draws: {scores['draw']}")

        again = input("\nPlay again? (y/n): ").strip().lower()
        if again != "y":
            print("Thanks for playing!")
            break

        change = input("Change difficulty? (y/n): ").strip().lower()
        if change == "y":
            difficulty = choose_difficulty()


if __name__ == "__main__":
    main()
```
