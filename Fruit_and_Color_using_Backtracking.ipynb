```python
# Fruit and Color using Backtracking

fruits = [
    ("apple", "red"),
    ("banana", "yellow"),
    ("orange", "orange"),
    ("grapes", "purple"),
    ("mango", "yellow"),
    ("watermelon", "green"),
    ("strawberry", "red")
]

color = input("Enter color: ").lower()

found = False

for fruit, fruit_color in fruits:
    if fruit_color == color:
        print("Fruit =", fruit)
        found = True

if not found:
    print("No fruit found.")
```
