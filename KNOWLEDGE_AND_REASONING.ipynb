KNOWLEDGE AND REASONING


```python
# Knowledge Base
weather = "rainy"

# Reasoning
if weather == "rainy":
    print("Knowledge: It is rainy.")
    print("Reasoning: Carry an umbrella.")
else:
    print("Knowledge: Weather is clear.")
    print("Reasoning: No umbrella needed.")
```

```python
# Knowledge Base
attendance = 80
marks = 65

# Reasoning
if attendance >= 75 and marks >= 50:
    print("Knowledge: Attendance and marks satisfy the rules.")
    print("Reasoning: Student is eligible for the exam.")
else:
    print("Knowledge: Requirements are not met.")
    print("Reasoning: Student is not eligible.")
```
