```python
# Graph
graph = {
    'A': [('B', 1), ('C', 4)],
    'B': [('D', 2)],
    'C': [('D', 1)],
    'D': []
}

# Heuristic values
h = {
    'A': 3,
    'B': 2,
    'C': 1,
    'D': 0
}

start = 'A'
goal = 'D'

open_list = [(start, 0)]  # (node, g_cost)
visited = []

while open_list:

    # Select node with minimum f = g + h
    current, g = min(open_list, key=lambda x: x[1] + h[x[0]])
    open_list.remove((current, g))

    print("Visiting:", current)

    if current == goal:
        print("Goal Reached!")
        break

    visited.append(current)

    for neighbor, cost in graph[current]:
        if neighbor not in visited:
            open_list.append((neighbor, g + cost))
```

```python
rooms = ["Room1", "Room2", "Room3", "Room4", "Room5"]
subjects = ["AI", "DBMS", "OS"]

conflicts = {
    "Room1": ["Room2", "Room3"],
    "Room2": ["Room1", "Room3", "Room4"],
    "Room3": ["Room1", "Room2", "Room5"],
    "Room4": ["Room2", "Room5"],
    "Room5": ["Room3", "Room4"],
}

def is_valid(room, subject, assignment):
    for n in conflicts.get(room, []):
        if assignment.get(n) == subject:
            return False
    return True

def backtrack(assignment):
    if len(assignment) == len(rooms):
        return assignment
    room = [r for r in rooms if r not in assignment][0]
    for subject in subjects:
        if is_valid(room, subject, assignment):
            assignment[room] = subject
            result = backtrack(assignment)
            if result:
                return result
            del assignment[room]
    return None

allocation = backtrack({})
for room in rooms:
    print(room, "->", allocation[room])
```
