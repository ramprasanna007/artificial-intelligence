```python
# Bird Flying System

birds = {
    "eagle": True,
    "sparrow": True,
    "parrot": True,
    "pigeon": True,
    "penguin": False,
    "ostrich": False,
    "emu": False
}

def can_fly(bird):

    if bird in birds:
        if birds[bird]:
            print(bird, "can fly.")
        else:
            print(bird, "cannot fly.")
    else:
        print("Bird not found.")


# Queries
can_fly("eagle")
can_fly("penguin")
can_fly("ostrich")
```
