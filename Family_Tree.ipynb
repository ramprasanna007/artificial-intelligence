```python
# Family Tree

parents = {
    "john": ["mary", "peter"],
    "mary": ["alice", "bob"],
    "peter": ["charlie"]
}

def parent(person, child):
    if person in parents and child in parents[person]:
        return True
    return False


def grandparent(person, child):

    if person in parents:
        for parent1 in parents[person]:
            if parent1 in parents and child in parents[parent1]:
                return True

    return False


def sibling(person1, person2):

    for parent_person, children in parents.items():
        if person1 in children and person2 in children:
            return True

    return False


# Queries

print("Is John parent of Mary?",
      parent("john", "mary"))

print("Is John grandparent of Alice?",
      grandparent("john", "alice"))

print("Are Alice and Bob siblings?",
      sibling("alice", "bob"))

print("Is Mary parent of Alice?",
      parent("mary", "alice"))
```
